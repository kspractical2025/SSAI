```python
# Creating side by side
np_arr = np.array([[1, 2, 3], [4, 5, 6]])
tf_tensor = tf.constant([[1, 2, 3], [4, 5, 6]])

print("NumPy shape:", np_arr.shape, "dtype:", np_arr.dtype)
print("TF shape:", tf_tensor.shape, "dtype:", tf_tensor.dtype)
```

```python
# Indexing & Slicing
print("\nIndexing (first row):")
print("NumPy:", np_arr[0])
print("TF:", tf_tensor[0])
```

```python
# Reshaping
print("\nReshaping to 3x2:")
print("NumPy:\n", np.reshape(np_arr, (3, 2)))
print("TF:\n", tf.reshape(tf_tensor, (3, 2)))
```

```python
# Mutability
np_arr[0, 0] = 99  # Allowed
print("\nMutated NumPy:", np_arr[0,0])
# tf_tensor[0, 0] = 99  # Not allowed for tf.constant
tf_var = tf.Variable(tf_tensor)
tf_var[0, 0].assign(99) # Allowed for tf.Variable
print("Mutated TF Variable:", tf_var[0,0].numpy())
```

```python
# 2. tf.Variable
var_tensor = tf.Variable([1, 2, 3])
var_tensor.assign([4, 5, 6])
var_tensor.assign_add([1, 1, 1])
print("\nvar_tensor:\n", var_tensor.numpy())
```

```python
zeros = tf.zeros((2, 3))
ones = tf.ones((2, 2))
print("\nzeros:\n", zeros)
print("ones:\n", ones)
```

```python
# 5. tf.reshape
flat = tf.reshape(zeros, [-1])
print("\nflattened zeros:", flat)
```

```python
# 6. tf.concat
t1 = tf.constant([[1, 2], [3, 4]])
t2 = tf.constant([[5, 6], [7, 8]])
concat_0 = tf.concat([t1, t2], axis=0) # 4x2
concat_1 = tf.concat([t1, t2], axis=1) # 2x4
print("\nconcat axis 0:\n", concat_0)
print("concat axis 1:\n", concat_1)
```

```python
# 7. tf.stack
stacked = tf.stack([t1, t2], axis=0) # 2x2x2
print("\nstacked:\n", stacked)
```

# Mathamatical Funcationhs

```python
a = tf.constant([[1, 2], [3, 4]])
b = tf.constant([[5, 6], [7, 8]])

print("Addition:\n", tf.add(a, b)) # or a + b
print("\nElement-wise Multiplication:\n", tf.multiply(a, b)) # or a * b
```

```python
# Matrix Multiplication
mat1 = tf.ones((3, 2))
mat2 = tf.ones((2, 4)) * 2
print("\nMatrix Multiplication (3x2 @ 2x4 = 3x4):\n", tf.matmul(mat1, mat2)) # or mat1 @ mat2

# Reduce functions
print("\nMean overall:", tf.reduce_mean(a))
print("Sum axis=0:", tf.reduce_sum(a, axis=0))
```
