```python
import pandas as pd
import numpy as np

# ---- Generate synthetic nominal data ----
np.random.seed(42)
categories = ['Red', 'Blue', 'Green', 'Yellow']
df = pd.DataFrame({
    'col': np.random.choice(categories, size=20, p=[0.4, 0.3, 0.2, 0.1])
})

print("Original data (first 5 rows):")
print(df.head())

# ---- One-Hot Encoding ----
one_hot = pd.get_dummies(df['col'], prefix='color', dtype=int)
df_ohe = pd.concat([df.drop(columns=['col']), one_hot], axis=1)

print("\nOne-Hot Encoded (first 5 rows):")
print(df_ohe.head())
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OrdinalEncoder

# ---- Generate synthetic ordinal data ----
np.random.seed(123)
levels = ['High School', 'Bachelor', 'Master', 'PhD']
# Create a meaningful distribution (e.g., more Bachelors)
df = pd.DataFrame({
    'col': np.random.choice(levels, size=20, p=[0.2, 0.4, 0.25, 0.15])
})

print("Original data (first 5 rows):")
print(df.head())

# ---- Ordinal Encoding ----
# Define the correct order (lowest to highest)
cat_order = [['High School', 'Bachelor', 'Master', 'PhD']]
encoder = OrdinalEncoder(categories=cat_order, dtype=int)
df['col_ordinal'] = encoder.fit_transform(df[['col']])

print("\nOrdinal Encoded (first 5 rows):")
print(df[['col', 'col_ordinal']].head())
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder

# ---- Generate synthetic binary data ----
np.random.seed(789)
df = pd.DataFrame({
    'col': np.random.choice(['Yes', 'No'], size=20, p=[0.6, 0.4])
})

print("Original data (first 5 rows):")
print(df.head())

# ---- Binary Encoding (0/1) ----
le = LabelEncoder()
df['col_binary'] = le.fit_transform(df['col'])

# Alternatively, manual mapping:
# df['col_binary'] = df['col'].map({'Yes': 1, 'No': 0})

print("\nBinary Encoded (first 5 rows):")
print(df[['col', 'col_binary']].head())
```
